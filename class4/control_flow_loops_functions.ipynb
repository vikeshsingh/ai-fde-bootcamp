```python
list_values = list(range(0,11))
for i in list_values:
    if i in [3,4,5]:
        pass
    else:
        print(i)
    print("Iteration completed")
```

```python
list_values = list(range(0,11))
for i in list_values:
    if i in [3,4,5]:
        continue # the execution will go back to for loop
    else:
        print(i)
    print("Iteration completed")
```

```python
list_values = list(range(0,11))
for i in list_values:
    if i == 3:
        break # the execution will come after the loop
    else:
        print(i)
    print("Iteration completed")

print("Loop completed.")
```

# Scope of a varible

```python
a = 10

def do_something():
    # local variable a
    a = 0
    a = a+1
    print(a)

do_something()
print(a)
```

```python
a = 10

def do_something():
    global a
    a = a+1
    print(a)

do_something()
print(a)
```

```python
a = 10

def do_something():
    print(a) # assignment and modition is not allowed
    # reading of global variable is allowed inside a function

do_something()
print(a)
```

```python
a = [1,2,3]

def do_something():
    a.append(123) # for collections modification is allowed
    print(a)

do_something()
print(a)
```
