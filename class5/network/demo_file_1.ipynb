# Generator

```python
list_a = range(1,100,3)

def get_list_values(list_temp):
    return list(list_temp)
```

```python
print(get_list_values(list_a))
```

```python
def get_list_values_generator(list_temp):
    for i in list_temp:
        yield i
```

```python
generator_a = get_list_values_generator(list_a)
```

# PDF READING DEMO Example

```python
def extract_all_pages(pdf_document):
    all_page_text = []

    for page in pdf_document:
        all_page_text.append(page.extract_text() or "")

    return all_page_text
```

```python
for i, value in enumerate(range(100,104), start=1):
    # index, value
    print(i, value)
```

```python
def extract_pages(pdf_document):
    for page_number, page in enumerate(pdf_document, start=1):
        page_text = page.ocr().extract_text() or ""
        yield  page_number, page_text
```

```python
pdf_fde_book = ""
pdf_generator = extract_pages(pdf_fde_book)
```

```python
next(pdf_generator)
```
