<h1>Operators</h1>

```python
num_1 = 50
num_2 = 20
print(num_1 + num_2)
print(num_1 - num_2)
print(num_1 * num_2)
#true division
print(num_1 / num_2)
#floor division
print(num_1 // num_2)
#moduler
print(num_1 % num_2)
#power
print(num_1 ** num_2)
```

```python
#comparison operator it will compare 2 values and return boolean
num_1 = 50
num_2 = 20
print(num_1 > num_2)
print(num_1 >= num_2)
print(num_1 == num_2)
print(num_1 <= num_2)
```

```python
#assignment
x = 10
x += 5
print(x)
x -= 5
print(x)
x *= 5
print(x)
x %= 3
print(x)
```

```python
#logical operators // and or not
a = 10
b = 20
c = 10
print(a is b)
print(a is c)
print(a is not c)
```

```python
# Bitwise Operator
a = 12
b = 4
print(a & b) # its basicaly created binary of both nubmer and perform & operator // see screenshot
print(a | b)
```
