<h1>Collection in Python</h1>

```python
#list
numbers = [2,5,8,1,4]
print(numbers)
print(type(numbers))
print(id(numbers))
numbers.append(1000) # add an item at end
print(numbers)
print(id(numbers))
numbers.append(1000)
print(numbers)
numbers.remove(1000) # only remove the first matching value in case of multiple value
print(numbers)
print(id(numbers))
numbers.pop(1) # this will remove value of index 1
print(numbers)
print(numbers[1])
numbers.sort()
print(numbers)
print(numbers.count(8))
numbers = [2,5,8,1,4]
print(id(numbers))
```

```python
#tuple // A tuple is a collection which is ordered and unchangeable.
thistuple = ("apple", "banana", "cherry")
print(thistuple)
print(id(thistuple))
#Tuples can also be created without the parentheses:
thistuple = "apple", "banana", "cherry"
print(thistuple)
print(id(thistuple))
# allow Duplicates
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print(thistuple)
print(id(thistuple))
```

```python
#Set items are unchangeable, but you can remove items and add new items.
# no duplicates if you store duplicates it will auto remove first one
thisset = {"apple", "banana", "cherry","apple"}
print(thisset)
thisset.add("mango") # this can be added any where no order
print(thisset)
```
