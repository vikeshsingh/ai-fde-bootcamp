<h1>Data Types in Python</h1>

```python
#int
a = 2
print(a)
print(type(a))
```

```python
#string
name = "Vikesh Singh"
print("Hello", name)
print(type(name))
```

```python
#float
num = 12.5675757575
print(num)
print(type(num))
```

```python
#bollean
isVerified = True
print(isVerified)
print(type(isVerified))
```

```python
#immutable object can’t be changed after it is created
#id() function prints numerical memory address of a variable
#Immutable Objects (e.g., int, float, bool, str, tuple)
a = 5
print(id(a))
a = 6
print(id(a))
a = 5
print(id(a))
```
